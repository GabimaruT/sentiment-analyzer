```python
import joblib
import re
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer

# Initialize NLTK utilities
ps = PorterStemmer()
stop_words = set(stopwords.words('english'))

# Wrap ALL your specific preprocessing logic into one function
def my_clean_processor(text):
    # 1. Remove HTML
    clean = re.compile('<.*?>')
    text = re.sub(clean, '', text)
    
    # 2. Lowercase
    text = text.lower()
    
    # 3. Clean special characters (your exact loop logic)
    x = ''
    for i in text:
        if i.isalnum():
            x += i
        else:
            x += ' '
            
    # 4 & 5. Tokenize, remove stopwords, and stem
    words = x.split()
    stemmed_tokens = [ps.stem(word) for word in words if word not in stop_words]
    
    # Return a list of cleaned tokens (which CountVectorizer expects from an analyzer)
    return stemmed_tokens

# Create the Pipeline
pipeline = Pipeline([
    ('vectorizer', CountVectorizer(analyzer=my_clean_processor)),
    ('classifier', clf2) # clf2 is your trained MultinomialNB model from the notebook
])

# Save the complete pipeline object
joblib.dump(pipeline, 'sentiment_pipeline.pkl')
```

```python
import re
import pandas as pd
import numpy as np
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
stop_words = set(stopwords.words('english'))

# 1. Remove HTML
def clean_data(text_list):
    clean = re.compile('<.*?>')
    # Use str(text) to ensure no float/NaN values break the regex
    return [re.sub(clean, '', str(text)) for text in text_list]

# 2. Lowercase
def convert_to_lowercase(text_list):
    return [text.lower() for text in text_list]

# 3. Remove Special Characters
def renmove_special(text_list):
    cleaned = []
    for text in text_list:
        x = ''.join([i if i.isalnum() else ' ' for i in text])
        cleaned.append(x)
    return cleaned

# 4. Remove Stopwords (splits into lists of words, removes stopwords)
def remove_stopwords(text_list):
    return [[word for word in text.split() if word not in stop_words] for text in text_list]

# 5. Stem Words (processes the list of words)
def stem_words(text_matrix):
    return [[ps.stem(word) for word in word_list] for word_list in text_matrix]

# 6. Join Back (CRITICAL: turns lists back into strings for CountVectorizer)
def join_back(text_matrix):
    # This ensures CountVectorizer receives standard text sentences
    return [' '.join(word_list) for word_list in text_matrix]
```
